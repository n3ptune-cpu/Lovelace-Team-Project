```python
import pandas as pd

# Read the CSV file
hd_data = pd.read_csv("heart.csv")
result=hd_data.loc[0]
#print(result)

stringified = str(hd_data)
splitted = stringified.split("\n")

data = []

for index in range(1, len(splitted), 1):
  raw_data = splitted[index].split(",")

  # Modify first row
  row_0 = str(raw_data[0])
  if len(row_0) <= 0:
    continue
  new_row_0 = row_0[len(row_0)-2::1]

  # Modify last row
  last_row = raw_data[len(raw_data)-1]
  new_last_row = last_row[0]

  # Apply modified data
  raw_data[len(raw_data)-1] = new_last_row
  raw_data[0] = new_row_0

  data.append(raw_data)


print(data)
```
