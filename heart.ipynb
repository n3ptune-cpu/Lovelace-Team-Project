```python
import pandas as pd
import plotly.offline as py
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go

# Read the CSV file
hd_data = pd.read_csv("heart.csv")


age = hd_data['age']
layout = go.Layout(barmode='overlay')
data = go.Histogram(x=age, opacity=0.6, xbins={'size': 4})
fig = go.Figure(data=[data], layout=layout)
py.offline.iplot(fig)
```

```python
import pandas as pd
import plotly.offline as py
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go

# Read the CSV file
hd_data = pd.read_csv("heart.csv")



hd_data['trestbps'].hist()
plt.title('Resting Blood pressure distribuition')
ax = sns.histplot(hd_data['trestbps'], bins=10, kde=True, label='Resting BP')

for patch in ax.patches:
    patch.set_alpha(0.4)

plt.legend()
plt.title('Resting Blood Pressure Density Plot')
plt.show()
```
