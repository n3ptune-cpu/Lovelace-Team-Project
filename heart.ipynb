```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the CSV file
hd_data = pd.read_csv("heart.csv")
#result=hd_data.loc[0]

sns.histplot(hd_data[hd_data['sex']==1]['age'], label='male', kde=True)
sns.histplot(hd_data[hd_data['sex']==0]['age'], label='female', kde=True)
plt.legend()
plt.title("Destiny plot of age by sex")
```
